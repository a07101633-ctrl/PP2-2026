# 12장 상속

- 상속에 대한 이론과 실제 학습

```python
class Car:
    def __init__(self, maker, model, color, price):
        self.maker = maker
        self.model = model
        self.color = color
        self.price = price

    def setMaker(self, maker):
        self.maker = maker

    def getMaker(self):
        return self.maker

    def getDesc(self):
        return self.__str__()
    def __str__(self):
        return f"{self.maker} {self.model} ({self.color}) - ${self.price}"

class ElectricCar(Car):
    def __init__(self, maker, model, color, price, bat_size):
        super().__init__(maker, model, color, price)
        self.batterySize = bat_size
    
    def getBatterySize(self):
        return self.batterySize

    def setBatterySize(self, bat_size):
        self.batterySize = bat_size

    def __str__(self):
        return f"{super().__str__()}, battery : {self.batterySize}kwh"

    def getDesc(self):
        return self.__str__()


class Truck(Car):
    def __init__(self, maker, model, color, price, payload):
        super().__init__(maker, model, color, price)
        self.payload = payload

    def __str__(self):
        return f"{super().__str__()}, payload : {self.payload}"


car1 = Car("Kia", "Ray", "Blue" ,2000)
print (car1.getDesc())
print (car1.model)

ecar1 = ElectricCar("Tesla", "Model S", "Red", 8000, 81)
print(ecar1)
ecar1.setBatterySize(92)
print(ecar1.getDesc())
truck1 = Truck("Ford", "F-150", "Black", 30000, 1000)
```

# 속성변수의 거시성: public, protected(_로 시작)vate(_로 시작)

```python
class Parent(object):
    def __init__(self):
        self.__private_money = 1000
        self._protected_money = 500
        self.public_money = 100


    def __str__(self):
        return f"Parent has total {self.__private_money + self._protected_money + self.public_money} money"

class child(Parent):
    def __init__(self):
        super().__init__()

    def __str__(self):
        return f"Child : {self.__private_money} private, {self._protected_money} protected, {self.public_money} public"

def test():
    p = Parent()
    print(p)
    c = child()
    print(c)

    print(p.public_money)
    print(p._protected_money)
    # print(p.__private_money)

    print(f"child : {c.public_money}")
    print(f"child : {c._protected_money}")
    # print(f"child : {c.__private_money}")  

test()

```

# 다중 상속 : 2개 이상의 부모클래스로부터 상속받을 때

```python
class Base1:
    def __init__(self, v1 = 10):
        self.attr1 = v1

    def display1(self):
        print(f"{self.attr1}")

    def display(self):
        self.display1()

class Base2:
    def __init__(self, v2 = 20):
        self.attr2 = v2

    def display2(self):
        print(f"{self.attr2}")

    def display(self):
        self.display2()

class C(Base1, Base2):
    def __init__(self, v1, v2, v3):
        Base1.__init__(self, v1)
        Base2.__init__(self, v2)
        self.attr3 = v3

    def print(self):
        print(f"{self.attr1}, {self.attr2}, {self.attr3}")
        self.display1()
        self.display2()
        self.display()
        Base1.display(self)
        Base2.display(self)


def test():
    c = C(v1=1, v2=2, v3=3)
    c.print()

    c2= C(4,5,6)
    c2.print()

if __name__=="__main__":
    test()
```

# Overriding : 부모의 함수(메소드) 재정의

```python
import math

class Shape:
    def __init__(self,name = "도형"):
        self.name = name

    def draw(self):
        print(f"Shape{self.name}을(를)그립니다.")

    def get_area(self):
        print(f"Shape{self.name}의 면적을 계산합니다")

class Circle(Shape):
    def __init__(self, radius=1,name="원"):
        super().__init__(name)
        self.radius = radius
    def draw(self):
        print(f"Circle{self.name}을(를)그립니다.")
        
    def get_area(self) -> float:
        print(f"Circle{self.name}의 면적을 계산합니다")
        return math.pi * self.radius ** 2
class Rectangle(Shape):
    def __init__(self, width = 1, height = 1, name = "사각형"):
        super().__init__(name)
        self.width = width
        self.height = height        


    
def test():
    c = Circle(10)  
    c.draw()
    print(c.get_area())

    r1= Rectangle(width = 5, height = 10)
    r1.draw()
    print(c.get_area())

    print("--------------------------------")
    circles = [Circle(1), Circle(2), Circle(3)]
    for c in circles:
        #c.draw()
        print(c.get_area())
    print("--------------------------------")

    print("--------------------------------")
    rectangles = [Circle(1), Circle(2), Circle(3)]
    for r in rectangles:
            #r.draw()
            print(r.get_area())
    print("--------------------------------")

    print("--------------------------------")
    shapes = [Circle(1), Rectangle(1,1), Circle(20), Rectangle(2,3)]
    for s in shapes:
        #s.draw()
        print(s.get_area())
    print("--------------------------------")


if __name__=="__main__":
    test()
```
