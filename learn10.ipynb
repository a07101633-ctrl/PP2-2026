# 10장 파일 및 예외처리

파일의 개념: 파일은 보조기억장치(영구저장장치) 에 기록된 논리적인 정보 단위이다.
파일 안에 저장된 정보는 바이트 단위로 순차적으로 기록되어 있다.
첫 번째 바이트의 위치를 오프셋(offset) 0라고 하고 1씩 증가한다.
파일을 읽고 쓰려면 먼저 파일을 열어야 한다.
열린 파일 포인터는 최초에 파일을 열 때는 0을 가리키며 접근한 바이트 수만큼 증가한다.
파일을 접근할 때는 열린 파일 포인터의 위치부터 읽거나 쓴다.
파일의 끝을 EOF (End of File) 라고 한다.

LAB 매출 파일 처리

```python
# 이 프로그램은 파일 안의 데이처를 읽어서 처리한다.
# 입력 파일 이름과 출력 파일 이름을 받는다.
infilename = input("입력 파일 이름을 입력하세요: ");
outfilename = input("출력 파일 이름을 입력하세요: ");

infile = open(infilename, "r")
outfile = open(outfilename, "w")
# 합계와 횟수를 위한 변수를 정의한다.
sum = 0
count = 0

# 입력 파일에서 한 줄을 읽어서 합계를 계산한다.
line = infile.readline()
while line != "" :
    s = int(line)
    sum =+ s
    count =+ 1
    line = infile.readline()

# 총매출과 일평균 매출을 출력 파일에 기록한다.
outfile.write("총매출 = "+ str(sum)+"\n")
outfile.write("평균 일매출 = "+ str(sum/count ))

infile.close()
outfile.close()

```

LAB 행맨

```python
# 이 프로그램은 행맨 게임을 구현한다.

import random

guesses = ''
turns = 10

infile = open("words.txt", "r")
lines = infile.readlines()
word = random.choice(lines)

while turns > 0:
    failed = 0
    for char in word:
        if char in guesses:
            print(char, end="")
        else:
            print("_", end="")
            failed += 1
    if failed == 0:
        print("사용자 승리")
        break
    print("")
    guess = input("단어를 추측하시오:")
    guesses += guess
    if guess not in word:
        turns -= 1 
        print ("틀렸음!")
        print (str(turns) + "기회가 남았음!")
        if turns == 0:
            print ("사용자 패배! 정답은 " + word)


infile.close()
```

LAB 각 문자 횟수 세기

```python
# 이 프로그램은 파일 안의 문자들의 개수를 센다.

filename = input("words.txt").strip()
infile = open("words.txt", "r")

freqs = {}

# 파일 각 줄에 대하여 문자를 추출한다. 각 문자를 사전에 추가한다.

for line in infile:
    for char in line.strip():
        if char in freqs:
            freqs[char] += 1
        else:
            freqs[char] = 1

print(freqs)
infile.close()
```
